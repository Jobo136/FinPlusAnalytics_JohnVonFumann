```python
# Para Behavioural

from src.config import create_spark_session, BEHAVIOURAL_PATH
from src.io import read_parquet
from src.profiling import basic_info, describe_df, report_missing, count_row_duplicates
from src.utils import get_numeric_columns
from src.outliers import mad_outliers_report, iqr_outliers_report
from src.cleaning import clean_behavioural

# Para Clients

from src.config import create_spark_session, CLIENTS_PATH, OUTPUT_DIR
from src.io import read_parquet, write_parquet, spark_to_csv
from src.profiling import basic_info, describe_df, report_missing, count_row_duplicates
from src.cleaning import clean_clients
```

```python
spark = create_spark_session("EDA_BEHAVIOURAL")

beh_df = read_parquet(spark, BEHAVIOURAL_PATH)

basic_info(beh_df, "BEHAVIOURAL")
describe_df(beh_df, "BEHAVIOURAL")
report_missing(beh_df, "BEHAVIOURAL")
count_row_duplicates(beh_df, "BEHAVIOURAL")

beh_numeric_cols = get_numeric_columns(beh_df)
mad_outliers_report(beh_df, beh_numeric_cols, "BEHAVIOURAL")
iqr_outliers_report(beh_df, beh_numeric_cols, "BEHAVIOURAL")

beh_clean = clean_behavioural(beh_df)

write_parquet(beh_clean, OUTPUT_DIR / "Behavioural_CLEAN")
spark_to_csv(beh_clean, OUTPUT_DIR / "Behavioural_CLEAN.csv")
```

```python
spark = create_spark_session("EDA_CLIENTS")

cli_df = read_parquet(spark, CLIENTS_PATH)

basic_info(cli_df, "CLIENTS")
describe_df(cli_df, "CLIENTS")
report_missing(cli_df, "CLIENTS")
count_row_duplicates(cli_df, "CLIENTS")

cli_clean = clean_clients(cli_df)

write_parquet(cli_clean, OUTPUT_DIR / "Clients_CLEAN")
spark_to_csv(cli_clean, OUTPUT_DIR / "Clients_CLEAN.csv")
```
