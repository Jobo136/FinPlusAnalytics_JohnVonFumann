```python
from pyspark.sql import SparkSession
import sys
sys.path.append('/home/jovyan/work/src') 

from csv_to_parquet import convert_csv_to_parquet_spark
```

```python
spark = (
    SparkSession.builder
    .appName("CsvToParquetPipeline")
    .master("local[*]")  # todos los cores disponibles en el contenedor
    .config("spark.sql.parquet.compression.codec", "snappy")
    .getOrCreate()
)
```

```python
# Nombre del CSV dentro de /data en el contenedor
input_csv_name = "BEHAVIOURAL.csv"          # <- cámbialo por el tuyo real
output_parquet_name = "BEHAVIOURAL" # carpeta de salida Parquet
data_dir = "/home/jovyan/work/data/"                    # volumen montado desde .../data/

use_header = True
infer_schema = True
single_file = True
repartition_num = 100
```

```python
parquet_path = convert_csv_to_parquet_spark(
    spark=spark,
    input_csv=input_csv_name,
    output_parquet=output_parquet_name,
    header=use_header,
    infer_schema=infer_schema,
    repartition=repartition_num,
    single_file=single_file,
    data_dir=data_dir,
)
```

```python
df_parquet = spark.read.parquet(parquet_path)
df_csv = spark.read.csv("/home/jovyan/work/data/BEHAVIOURAL.csv", header=True)
df_csv.show(5)
df_parquet.show(5)
df_parquet.printSchema()
```
